### Create a ChromaDB collection and add documents, ids, and metadatas

```python
collection = chroma_client.create_collection(name="test_collection")
```

```python
collection.add(
    ids=ids,
    documents=documents,
    metadatas=metadatas,
)
```

### Querry fom DB

```python
results = collection.query(
    query_texts=["database for storing AI embeddings"],
    n_results=3
)

results
```

### Querry with Filtering

```python
results = collection.query(
    query_texts=["intelligent creatures"],
    n_results=3,
    where={"category": "animals"}
)

results
```
